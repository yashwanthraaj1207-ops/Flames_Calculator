```python
%%writefile app.py
import streamlit as st

# 🎨 Background Switch Option
bg_option = st.selectbox(
    "Choose Background Style 💅",
    ["Vibrant Love 💘", "Romantic Pink 💕", "Dark Passion 🖤"]
)

if bg_option == "Vibrant Love 💘":
    bg_style = "linear-gradient(135deg, #ff0066, #ff4da6, #ff1a75)"
elif bg_option == "Romantic Pink 💕":
    bg_style = "linear-gradient(to right, #ff5f6d, #ffc371)"
else:
    bg_style = "linear-gradient(135deg, #1e1e2f, #3a0ca3, #7209b7)"

# Apply Background
st.markdown(
    f"""
    <style>
    .stApp {{
        background: {bg_style};
        background-attachment: fixed;
    }}
    html, body {{
        color: white !important;
    }}
    </style>
    """,
    unsafe_allow_html=True
)

# BACKEND

def flames_result(name1, name2):
    name1 = name1.replace(" ", "")
    name2 = name2.replace(" ", "")

    if not name1 or not name2:
        return "Names cannot be empty!"

    if not name1.isalpha() or not name2.isalpha():
        return "Invalid input! Use letters only."

    x = list(name1.lower())
    y = list(name2.lower())

    for ch in x[:]:
        if ch in y:
            x.remove(ch)
            y.remove(ch)

    total = len(x) + len(y)

    if total == 0:
        return "No Result"

    flames = [
        "Not much it's just FRIENDSHIP...",
        "Congratulations! for your LOVE...",
        "Just an AFFECTION...pch",
        "OMG!..Wanna Know About Your Marriage",
        "Be Aware Of Dogs Bro it's ENEMY...",
        "Protect them it's your SIBLING..."
    ]

    index = 0
    while len(flames) > 1:
        index = (index + total - 1) % len(flames)
        flames.pop(index)

    return flames[0]

# UI

st.title("🔥 FLAMES Love Calculator 🔥")

name1 = st.text_input("Enter Your Name")
name2 = st.text_input("Enter Your Crush Name")

if st.button("Check Relationship 💘"):
    result = flames_result(name1, name2)
    st.success(f"Result: {result}")
```
